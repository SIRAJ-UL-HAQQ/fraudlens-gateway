# FraudLens Gateway — Exploratory Data Analysis (EDA)
This notebook explores the credit card fraud dataset, inspects extreme class imbalance, and checks feature distributions.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from src.load_data import load_dataset

df = load_dataset()
print('Dataset shape:', df.shape)
print('Class balance:\n', df['Class'].value_counts(normalize=True))
```

```python
# Distribution of Transaction Amounts
plt.figure(figsize=(10, 4))
sns.histplot(df[df['Class'] == 0]['Amount'], bins=50, color='blue', alpha=0.5, label='Legit')
sns.histplot(df[df['Class'] == 1]['Amount'], bins=50, color='red', alpha=0.5, label='Fraud')
plt.yscale('log')
plt.legend()
plt.title('Log Transaction Amount by Class')
plt.show()
```
